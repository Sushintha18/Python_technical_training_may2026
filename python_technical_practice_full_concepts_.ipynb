```python
#36. CIRCULAR PRIME ( all rotation of number should be prime)

n = input()
for i in range(len(n)):
    r = n[i:] + n[:i]
    num = int(r)
    c = 0
    for j in range(1, num + 1):
        if num % j == 0:
            c = c + 1
    if c == 2:
        print(r, "Prime")
    else:
        print(r, "Not Prime")
```

```python
#37 #1. scenario based problems:
'''1. AI Message Recovery System
Quantum Communication Center
A quantum communication satellite receives encrypted messages from deep-space probes.
Due to signal corruption, one character in a message may be incorrect.
Scientists need to determine whether the message can become a perfect mirror pattern (palindrome) by removing exactly one character.
If possible, print "YES", otherwise print "NO".
Input Format
Single string s
Constraints
•	1 <= length of s <= 10^5
Output Format
Print "YES" or "NO".
Sample Input
abca
Sample Output
YES
Explanation
Removing 'b' produces "aca", which is a palindrome.
'''


s=input()
count=0
for i in range(len(s)):
  st=s[:i]+s[i+1:]
  if(st==st[::-1]):
    print("YES")
    break
  else:
    count+=1
if(count==len(s)):
  print("NO")
else:
  pass
```

```python
#40
'''4. Trending Keyword Analyzer
Social Media Intelligence Engine
A social media company tracks millions of hashtags daily.
The analytics engine must identify the most frequently used hashtag.
If multiple hashtags have the same frequency, the system should choose the alphabetically smallest hashtag.

Input Format
Single line containing hashtags separated by spaces.
Sample Input
apple banana apple banana cat
Sample Output
apple
Explanation
Both "apple" and "banana" appear twice.
Alphabetically, "apple" comes first.
'''

num=list(input().split())
num.sort()
itemdict={}
for n in num:
  itemdict[n]=num.count(n)
values=list(itemdict.values())
keys=list(itemdict.keys())
for i in range(len(values)):
  if(values[i]==max(values)):
    print(keys[i])
    break
print(itemdict)
```

```python
#44
'''8.Aircraft Collision Warning System
Air Traffic Control Network
Two aircraft positions are represented as coordinate tuples.
The warning system calculates the distance between them.
If the distance is less than 10 km, print "Collision Risk".
Otherwise, print "Safe".

Input Format
First aircraft coordinates (x1, y1)
Second aircraft coordinates (x2, y2)
Sample Input
(2,3)
(5,7)
Sample Output
Collision Risk
'''


x1, y1 = eval(input())
x2, y2 = eval(input())
d = (((x2 - x1) ** 2 + (y2 - y1) ** 2))**(1/2)
if d < 10:
    print("Collision Risk")
else:
    print("Safe")
```
